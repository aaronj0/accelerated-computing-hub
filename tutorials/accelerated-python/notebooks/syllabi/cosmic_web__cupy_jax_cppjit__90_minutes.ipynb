### Introduction

This ninety-minute block computes one quantity four ways. The quantity is the Euler characteristic curve of a cosmic-web density field: threshold the field, count blobs minus tunnels plus cavities, track the count against the threshold. The intro builds that count from pixels up, checks it against the closed form for a Gaussian field, and times the CuPy reference. Four notebooks then take the same signed histogram through CuPy's trace, JAX, cuda.compute's iterators and a CUDA kernel written in the session, each checked against the reference by exact integer equality, and the synthesis compares what each programming model decided for you and what it cost.
